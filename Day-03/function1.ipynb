```python
# def : a function is a block of reusable code that perform  a specific task

# define the function 

def fun():
    print("welcome to functions")

print("Hello World")

fun()      # calling the bfunction
```

```python
# passing two parameter in the function

def fun(val1, val2):
    print("welcome to functions", val1)
    print("Hello World", val2)

fun(100, 200)      
```

```python
# passing one parameter

def fun(val1):
    print("welcome to functions", val1)
    print("Hello World")

fun(100)      
```
