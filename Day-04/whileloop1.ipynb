```python
# example 6: nested loop
# for every value of outerloop, inner loop will completly excute
for i in range(1,4):
    for j in range(1,3):
        print(i,j)
```

```python
for i in range(3,6):
    for j in range(2,5):
        print(i,j)
```

```python
for i in range(3,6): # n = 3 times
    for j in range(2,7): # m = 5 times
        print(i,j)
```

```python
# example 7: else satement
# else statement excutes if loop complete excution sucessfully
for i in range(5):
    if i == 2:
        continue
    elif i == 4:
         break
    else:
        print(i)
    pass
else:
    print("loop completed sucessfully")
```
