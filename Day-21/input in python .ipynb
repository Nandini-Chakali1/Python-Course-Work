```python
# 1. concatenation using commas
Hero_name = 'Prabhas'
age = 40
fee = 100
print(Hero_name,age,fee)
```

```python
# 2. string concatenation using +
Hero_name = 'Prabhas'
Herion_name = 'Anushka'
fee = 100
print(Hero_name + " " + Herion_name + " " +str(fee))
```
