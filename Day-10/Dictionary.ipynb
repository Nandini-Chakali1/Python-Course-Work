```python
empty_dict = {}
empty_dict2 = dict()
print(empty_dict)
print(type(empty_dict2))

int_dict = {1: 'Mamidipandu', 2 : 'Aratipandu', 3: 'Batai'}
print(int_dict)


float_dict = {1.2: 'Mamidipandu', 2.7 : 'Aratipandu', 3.7: 'Batai'}
print(float_dict)


str_dict = {"A": 'Mamidipandu', "B" : 'Aratipandu', "C": 'Batai'}
print(str_dict)


tuple_dict = {(10,20): 'Mamidipandu', 20.5 : 'Aratipandu', "c": 'Batai'}
print(tuple_dict)

het_dict = {10: 'Mamidipandu' , 20.5 : 'Aratipandu', "c": 'Batai'}
print(het_dict)

# tuple  can be used as a key
# list and set cannot be used as a key
# keys must be immutable
# values can be mutable and immutable

dict_comp = {x : x**2 for x in range(1,6)}
print(dict_comp)
```

## Accessing dictionary items

```python
drink_items = { 
                 "Thumpsup" : 50,
                 "Sugarcane" : 20,
                 "Coconutwater" : 60,
                 "Water" : 20,
                 "rosemilk" : 30
}


val = drink_items["Sugarcane"]
print(val)

val2 = drink_items.get("water")
print(val)
    
```

```python
val2 = drink_items.get("Softdrink")
print(val2)
    
```

```python
for i in drink_items:
   print(i, drink_items[i])
    
```

```python
print(drink_items.keys())
```

```python
print(drink_items.values())
```

```python
print(drink_items.items())
```

```python
# nested dictionary

nested_dict = {
                 100 : {"Hello" : 10, "World" : 20},
                 200 : {"A" : 50, "B" : 60}
}

nested_dict[100]["World"]
```
