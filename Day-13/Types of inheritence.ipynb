## types of inheritance

## single inheritance

```python
# parent class
class hyd_food:
    taste = "very very spicy"
    def biryani(self):
        print("very very tasty biryani")

    # child class
class Pistahouse(hyd_food):
    pass
    
    
obj_child = Pistahouse()
print(obj_child.taste)
```

```python
## example 2 super() function
# super(): used acess the parent properties and function from inside of the class

class hyd_food:
    taste = "very very spicy"
    def biryani(self):
        print("very very tasty biryani")

    # child class
class Pistahouse(hyd_food): 
    def display(self):
        super().biryani()
        print(super().taste)
        
obj_child = Pistahouse()
obj_child.display()
```

```python
## example 2 super() function
# super(): used acess the parent properties and function from inside of the class

class hyd_food:
    taste = "very very spicy"
    def biryani(self):
        print("very very tasty biryani")

    # child class
class Pistahouse(hyd_food): 
    def display(self):
        super().biryani()
        print(super().taste)
        
obj_child = Pistahouse()
obj_child.display()
```

## multiple inheritance

```python
## example 2
class father:
    def fun1(self):
        print(" i am father")
class mother:
    def fun2(self):
        print("i am mother")
class child(father, mother):
    def fun3(self):
        print("i am child")

obj1 = child()
obj1.fun1()
obj1.fun2()
obj1.fun3()
```

```python
## example 3    constructor function

class A:
    def __init__(self):
        print("i am a class constructor")

class B:
    def __init__(self):
        print(" i am B class constructor")
class C:
    def __init__(self):
        print("i am C class constructor")

obj_C = C()
    
```

```python
## super()  function

class A:
    def display(self):
        print("i am a class constructor")

class B:
    def show(self):
        print(" i am B class constructor")
class C(A, B):
    def hello(self):
        super().display()
        super().show()

obj_C = C()
obj_C.hello()    
```

# Multilevel Inheritance

```python
# Example-1: Structure of the multi inheritance
class grandGF:
    def Gold(self):
        print("1000 KG")

class grandfather(grandGF):
    def Land(self):
        print("500 Acr")

class father(grandfather):
    def Car(self):
        print("Lamborgini")

class son(father):
    def Bike(self):
        print("RE bike")
        
obj_son = son()
obj_son.Bike()
obj_son.Car()
obj_son.Land()
obj_son.Gold()
```

# Hybrid inheritance

```python
# example-1: structure of hybrid inheritance
class A:
    A_var = "A Class"

class B(A):
    B_var = "B Class"

class C(B):
    C_var = "C Class" 

class D(B):
    D_var = "D Class"

class E(C,D):
    E_var = "E Class"

obj_E = E()
print(obj_E.A_var)
```
