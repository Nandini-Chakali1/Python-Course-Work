```python
## the word polymorpohisam  means "many forms"


## in polymorphisam same function or same operator can perform many different task.

## types of polymorphisam:

# 1: function overloading
# 2: operator overloading
# 3: method overiding
```

```python
def fun(var1, var2 = None):
    if var2 is not None:
        print(var1, var2)
    
    else:
        print(var1)
fun(10)
fun(10, 20)
```

```python
## example 2 : default arguments 

def calculator(a = 0, b = 0, c = 0, d = 0):
    return a + b + c + d
print(calculator(100, 200))
print(calculator(100, 200, 300,))
print(calculator(100, 200, 300, 400))
```
