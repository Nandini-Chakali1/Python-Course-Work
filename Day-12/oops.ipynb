# encapsulation

```python
# encapusaltion refers to the binding of data (attributes) and functions into one unit(class)

# encapusaltion restrict the direct access of the data
```

```python
class Bike:
    No_tyre = 2 # class varible
    
    def __init__(self, name, price):     # constructor
        self.bike_name = name          # instanace varible
        self.price = price
        
    def display(self):
        
        print(self.No_tyre,self.bike_name,self.price)
        
obj1 = Bike("pulser", 1.3)
obj1.display()

print(obj1.No_tyre)
print(obj1.bike_name)
print(obj1.price)
        
```

##access specifier / types of attribute in encapsulation

```python
class marvels:
    def __init__(self, name, power, weapon): # arguments
       self.Name = name       #    public instance varible
       self._Power = power   # protected instance varible
       self.__Ayudham = weapon  # private
    
    def show(self):
        print(self.Name, self._Power, self.__Ayudham)
    
obj1 = marvels("Thor", 1000, "Hammer")
obj1.show()

print(obj1.Name)
print(obj1._Power)   # public 

## protected varibles can be access outside the class but you should not acess
# in python public, protected private concept is used just for sign/warning

print(obj1._marvels__Ayudham)   # obj._class_private varible
```
