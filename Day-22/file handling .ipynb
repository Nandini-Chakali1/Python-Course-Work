```python
## to write the content inside the file
## we will use the "w" mode the write() function
## if file don't exist then "w" mode will create a new file
## read mode "r" don't create the file
```

```python
with open("nandini_123.txt", "w") as f:
    f.write(" i am nandini")
```

```python
with open("nandini_123.txt", "r") as f:
    print(f.read())
```

```python
## IN "w" mode data get overwrite on previous data
with open("nandini_123.txt", "w") as f:
    f.write("my best friend mallika")
```

```python
with open("nandini_123.txt", "r") as f:
    print(f.read())
```

## append mode: a

```python
with open("nandini_123.txt", "a") as f:
    f.write("priya is very innocent")
```

```python
with open("nandini_123.txt", "r") as f:
    print(f.read())
```

```python
## Writeliness: writing multiple lines together in the file
```

```python
## w+ mode : write/read mode
# w+ mode will clear the previous content
# w+ create new file if file don't exist

with open("mallika_123.txt", "w+") as f:
    f.write("this is file handling")
    f.seek(0)
    print(f.read())
```

```python
with open("mallika_123.txt", "w+") as f:
    f.write("this is file handling")
    f.seek(6)
    print(f.read())
```

# r+ : read/write mode

```python
## r+ read/write
## file must exist ,otherwise shows error

with open("mallika_123.txt", "r+") as f:
    f.write("this is file handling")
    f.seek(0)
    print(f.read())
```

```python
## a+ : append/read
with open("mallika_123.txt", "a+") as f:
    f.write("this is file handling")
    f.seek(0)
    print(f.read())
```
