```python
## Bitwise operator (<<,>>,~, &,|,^)
```

```python
# bitwise operators works on binary number
# bitwise  operators faster than the arthimetic operators
# bitwise operators perform many operations in easy way then due to binary number system
```

```python
a = 10
b = 3
print(a & b)
print(a | b)
print(a ^ b)
      
```

```python
a = 10
b = 3
c = 2
print(a & b & c)
```

```python
print(a | b | c)
```

```python
## left shift (<<), right shift(>>)
```

```python
# def ( lef shift): insert new bits (0) to the lsb  lsb double the number
# def ( right shift): delete the bits from the lsb            formula a<< k   # 
```

```python
a  = 10 
print(a << 1) # left shift by 1 bit
```

```python
print(a << 2) # left shift by 2 bit
```

```python
a = 80
print(a << 6) # left shift by 4 bit
```

```python
## right shift (>>) in right shift the number will be half
```

```python
a = 64     deleteing bits from lsb
print(a >> 1)
```

```python
a = 64 
print(a >> 2)
```

 # comparison operators(==,>=, <=, !=, <, >)

```python
a = 10
b = 5
print(a == b)
print(a >= b)
print(a <= b)
print(a != b)
print(a < b)
```

## logical operators ( and , or not)

```python
a = True
b = False
print(a and b)
print(a or b)
print(not b)
 
```

```python
# identity operators (is is not)
a = [1,2,3]
b = [1,2,3]
c = a
print(a == b)
print(a is b)
print(a is c)
print(a is not c)
```
