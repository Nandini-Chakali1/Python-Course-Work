```python
# example 1  structure of class
# class is keyword, food is the class name

# class varible value cannot be change by object
# class varible value can be only change by class name

class food:
    food_name = "biryani"  # class varible

obj1 = food() # creating the object (obj) ofn food class
print(obj1.food_name) # accesing the class varible

obj2 = food()
print(obj2.food_name)    

# acess the class varible with class name 

print(food.food_name)
```

```python
class food:
    food_name = "biryani"  # class varible

obj1 = food() # creating the object (obj) ofn food class
print(obj1.food_name) # accesing the class varible

# class varible value can be only change by class name

food.food_name = "mutton"

obj2 = food()
print(obj2.food_name)    

# acess the class varible with class name 

print(food.food_name)

# class varible value cannot be change by object

obj3 = food()
obj3.food_name = "pappu"   # obj3 is creating its own local varible with same name

print(obj3.food_name)  # printing local varible to obj3
```

```python
## example 2: Constructor function
## constructor function automatically get called when object is created
## construction function used for initializing the varibles whikle creating the new object

## instance varible value is different for every object 
# class varible value  is same  for all objects
# self is a varible which points to the currents object  

class Bike:
    No_tyre = 2 # class varible
    
    def __init__(self):  #  constructor 
        print("i am constructor function")
    
obj1 = Bike()
obj1 = Bike()
```

```python
class Bike:
    No_tyre = 2 # class varible
    
    def __init__(self, name): # constructor
        self.bike_name = name   # instanace varible
        print(self.bike_name)
    
obj1 = Bike("royal enfiled")
obj1 = Bike("pulser")
```

```python
### example 3: member functions

# member function don't get called automatically, we need to call them ourself

class Bike:
    No_tyre = 2 # class varible
    
    def __init__(self, name, price): # constructor
        self.bike_name = name
        self.price = price

        print(self.bike_name)
        print(self.price)

    def display(self):  
        print("i am member function")
    
obj1 = Bike("royal enfiled", 1.2)
obj2 = Bike("pulser", 3)
obj1.display()
obj2.display()


obj1.bike_name    # instance varible calling outside the class 
```
