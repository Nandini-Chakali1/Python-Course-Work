## creating differnt tyoe of tuple

```python
empty_tuple = ()
empty_tuple2 = tuple() # creating empty tuple we have to use tuple function

print(empty_tuple, type(empty_tuple))

print(empty_tuple2, type(empty_tuple2))
```

```python
het_tuple = (10, 10.3, "hello")
print(het_tuple)
```

```python
dup_tuple = (10, 10 , 10, 20, 30)
print(dup_tuple)
```

```python
nested_tuple = ((10, 20, 30,) , [40, 50], {60, 70, 80, 90}, "Hello")
print(nested_tuple)
```

```python
# convert list and set to tuple ()
list1 = [10, 20, 30, 40]
tuple1 = tuple(list1)
print(tuple1)
```

```python
set1 = {10, 20, 30, 40}
tuple2 = tuple(set1)
print(tuple2)
```

```python
print(tuple(range(1,11)))
```

```python
t1 = (1,2,3)
t2 = (4,5,6)
t3 = (7,8,9)
t4 = t1 + t2 + t3
print(t4)
```

```python
t5 = tuple(i for i in range(10))
print(t5)
```

```python
# insert
tuple3 = (10, 20, 30)
list1 = list(tuple3)
print(list1)
list1.append(500)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
# delete
tuple3 = (10, 20, 30)
list1 = list(tuple3)
print(list1)
list1.remove(30)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```
